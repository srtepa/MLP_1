Имплементировать калькулятор с логированием операций:
- Обработка ошибок деления на ноль, неверного ввода, попытки логарифмирования отрицательного числа.
- Логировать все действия, в том числе ошибочные, в файл calc.log.

```python
import logging
import math

logging.basicConfig(
    level=logging.INFO,
    filename='calc.log',
    filemode='a',
    format='%(asctime)s - %(levelname)s - %(message)s',
    encoding='utf-8'
)

OPERATIONS = [
    "1. Сложение",
    "2. Вычитание",
    "3. Умножение",
    "4. Деление",
    "5. Возведение в степень",
    "6. Десятичный логарифм (только для первого числа)"
]

def calculate():
    print("Калькулятор")

    try:
        line = input("Введите два числа через пробел: ")
        a_raw, b_raw = line.split()
        a, b = float(a_raw), float(b_raw)
        logging.info(f"Введены числа: {a}, {b}")

        for op in OPERATIONS:
            print(op)

        choice = input("Выберите номер операции: ")
        logging.info(f"Выбрана операция: {choice}")

        result = None
        match choice:
            case "1":
                result = a + b
                logging.info(f"Сумма: {result}")
            case "2":
                result = a - b
                logging.info(f"Разность: {result}")
            case "3":
                result = a * b
                logging.info(f"Умножение: {result}")
            case "4":
                if b == 0:
                    logging.error(f"ОШИБКА: Деление на ноль!")
                    raise ZeroDivisionError("Деление на ноль!")
                result = a / b
                logging.info(f"Деление: {result}")
            case "5":
                result = a ** b
                logging.info(f"Возведение в степень: {result}")
            case "6":
                if a <= 0:
                    logging.error(f"ОШИБКА: Логарифм отрицательного числа или нуля!")
                    raise ValueError("Логарифм отрицательного числа или нуля!")
                result = math.log10(a)
            case _:
                print("Ошибка: Неверный выбор операции.")
                logging.warning(f"Некорректный выбор операции: {choice}")
                return

        if result is not None:
            print(f"Результат: {result}")
            logging.info(f"Результат: {result}")

    except ValueError as e:
        error_msg = f"Ошибка ввода или значения: {e}"
        raise ValueError(error_msg)
        logging.error(error_msg)
    except ZeroDivisionError as e:
        error_msg = f"Математическая ошибка: {e}"
        raise ZeroDivisionError(error_msg)
        logging.error(error_msg)
    except Exception as e:
        error_msg = f"Непредвиденная ошибка: {e}"
        raise Exception(error_msg)
        logging.exception("Critical Error")

if __name__ == "__main__":
    calculate()
```
