```python
from math import log, exp
import numpy as np
import time
import matplotlib.pyplot as plt

sizes = [10**4, 10**5, 10**6, 10**7]

op = ["+", "*", "log", "exp"]

results_np = {i: [] for i in op}
results_py = {i: [] for i in op}
```

```python
for i in sizes:
    np_arr = np.arange(1, i + 1)
    py_list = [j for j in range(1, i+1)]

    #numpy
    start = time.time()
    _ = np_arr + 1
    results_np["+"].append(time.time() - start)

    start = time.time()
    _ = np_arr * 4
    results_np["*"].append(time.time() - start)

    start = time.time()
    _ = np.log(np_arr)
    results_np["log"].append(time.time() - start)

    start = time.time()
    _ = np.exp(np_arr / i)
    results_np["exp"].append(time.time() - start)

    #python list
    start = time.time()
    _ = [x + 2 for x in py_list]
    results_py["+"].append(time.time() - start)

    start = time.time()
    _ = [x * 4 for x in py_list]
    results_py["*"].append(time.time() - start)

    start = time.time()
    _ = [log(x) for x in py_list]
    results_py["log"].append(time.time() - start)

    start = time.time()
    _ = [exp(x / i) for x in py_list]
    results_py["exp"].append(time.time() - start)
```

```python
for operation in op:
    plt.plot(sizes, results_np[operation], label='NumPy')
    plt.plot(sizes, results_py[operation], label='Python list')

    plt.title(f"Операция {operation}")
    plt.xlabel("Размер массива")
    plt.ylabel("Секунды")
    plt.legend()

    plt.show()
```
